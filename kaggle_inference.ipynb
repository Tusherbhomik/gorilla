```python
# Step 1: Clone the repo
!git clone https://github.com/Tusherbhomik/gorilla.git /kaggle/working/gorilla
```

```python
# Step 2: Navigate to BFCL directory
%cd /kaggle/working/gorilla/berkeley-function-call-leaderboard
```

```python
# Step 3: Install BFCL package + remove flashinfer (use Kaggle's pre-installed vllm, don't reinstall it)
!pip install -q -e .
!pip uninstall -y flashinfer flashinfer-python 2>/dev/null || true
print("Done. flashinfer removed — vLLM will use TRITON_ATTN.")
```

```python
# Step 4: Set Triton attention backend (flashinfer is gone, this is the fallback)
import os
os.environ['VLLM_ATTENTION_BACKEND'] = 'TRITON_ATTN'
print('Backend:', os.environ['VLLM_ATTENTION_BACKEND'])
```

```python
# Step 5: Generate (inference)
!VLLM_ATTENTION_BACKEND=TRITON_ATTN python -m bfcl_eval generate \
    --model tusherbhomik/qwen2.5-1.5b-hgr-5340-r2 \
    --test-category non_live,live,multi_turn \
    --num-gpus 1 \
    --backend vllm \
    --allow-overwrite
```

```python
# Step 6: Evaluate
!python -m bfcl_eval evaluate \
    --model tusherbhomik/qwen2.5-1.5b-hgr-5340-r2 \
    --test-category non_live,live,multi_turn
```

```python
# Step 7: Show scores
!python -m bfcl_eval scores
```
